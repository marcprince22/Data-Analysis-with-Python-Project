```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw_data/socioeconomic_factors/time_series_socio_factors/education.csv", sep=";")
```

```python
print(df.shape)
df.head()
```

```python
df = df[df["GDENR"] <= 298]

print(df.shape)
print(df["GDENR"].nunique())
```

```python
df = df[["GDENR", "ERHJAHR", "TERTIARY"]]
#keep relevant columns
```

```python
df = df.rename(columns={"ERHJAHR": "year","TERTIARY": "tertiary_share"})
#rename that stuff
```

```python
df.head()
```

```python
print(df.dtypes)
print(df.isna().sum())
print(df.duplicated(subset=["GDENR", "Jahr"]).sum())
print(df["tertiary_share"].describe())
print(df.groupby("Jahr")["GDENR"].nunique())
```

```python
df.to_csv("../data/clean_data/ts_tertiary_share_clean.csv", index=False)
```
