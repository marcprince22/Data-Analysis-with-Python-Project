```python
import pandas as pd

turnout = pd.read_csv("../data/clean_data/ts_zh_turnout_clean.csv")
socio = pd.read_csv("../data/clean_data/ts_zh_socioeconomic_panel_clean.csv")
```

```python
print(turnout.shape)
print(socio.shape)
print(turnout.dtypes)
print(socio.dtypes)
```

```python
socio.sample(20)
```

```python
panel = turnout.merge(socio, on=["GDENR", "year"], how="inner")
#merge - just keep the rows which find a partner
```

```python
print(panel.shape)
print(panel["GDENR"].nunique())
print(panel["year"].min(), panel["year"].max())
print(panel.duplicated(subset=["GDENR", "year"]).sum())
```

```python
panel.head()
```

```python
panel = panel.sort_values(["GDENR", "year"])
panel.to_csv("../data/clean_data/zh_regression_panel.csv", index=False)
```
